The dataset student_dropout_risk_2026.csv.gz contains one row for each university student (identified by person_id), with demographic, economic, academic and behavioural information. The column dropout is 1 if the student has dropped out, 0 otherwise. All columns whose values are 0/1 are boolean flags.

1- Extract all students that study Online, have an attendance_rate smaller than 50, and have more than 5 missed_deadlines. Store them in a dataframe called at_risk.

```python
import pandas as pd
```

```python
df = pd.read_csv('student_dropout_risk_2026.csv.gz')
df.head()
```

```python
df.columns
```

```python
at_risk = df[(df['study_mode'] == 'Online') & (df['attendance_rate'] < 50) & (df['missed_deadlines'] > 5)]
at_risk
```

2- Then compute, for each course_type, how many of those students there are and what fraction of them has dropped out. Store the result in a dataframe called at_risk_by_course, with columns students and dropout_rate, sorted by decreasing dropout_rate.

```python
students = at_risk.groupby('course_type')['person_id'].count()
dropout_rate = at_risk.groupby('course_type')['dropout'].mean()
at_risk_by_course = pd.DataFrame({'students' : students, 'dropout_rate' : dropout_rate}).sort_values(by = 'dropout_rate', ascending = False)
at_risk_by_course
```

3- Add to the dataframe the following columns:

pass_rate: the ratio between units_passed and units_enrolled;
gpa_change: the difference between gpa_semester_2 and gpa_semester_1;
gpa_trend: a string that is "improving" if gpa_change is larger than 0.5, "declining" if gpa_change is smaller than -0.5, and "stable" otherwise.

```python
df['pass_rate'] = df['units_passed'] / df['units_enrolled']
df['gpa_change'] = df['gpa_semester_2'] - df['gpa_semester_1']
df['gpa_trend'] = 'stable'
df.loc[df['gpa_change'] > 0.5, 'gpa_trend'] = 'improving'
df.loc[df['gpa_change'] < -0.5, 'gpa_trend'] = 'declining'
df
```

4- Verify that, for every student, units_passed + units_failed equals units_enrolled; if not, report the rows violating this condition.

```python
violating = df[df['units_enrolled'] != (df['units_passed'] + df['units_failed'])]
if len(violating) == 0:
    print ('No student violates the condition')
else:
    print(violating)
```

5- Compute the dropout rate for each value of gpa_trend, and the person_id of the 10 students with the largest GPA drop.

```python
df.groupby('gpa_trend')['dropout'].mean()
```

```python
lim = df['gpa_change'].sort_values().iloc[9]
selected = df[df['gpa_change'] <= lim].sort_values('gpa_change')
selected[['person_id', 'gpa_change']]
```

6- Partition the students into the following groups, according to weekly_work_hours: 0, 1-10, 11-20, 21-30, more than 30 (all intervals include both extremes).

```python
df.loc[df['weekly_work_hours'] == 0, 'wwh_group'] = '0'
df.loc[(df['weekly_work_hours'] >= 1) & (df['weekly_work_hours'] <= 10), 'wwh_group'] = '1-10'
df.loc[(df['weekly_work_hours'] >= 11) & (df['weekly_work_hours'] <= 20), 'wwh_group'] = '11-20'
df.loc[(df['weekly_work_hours'] >= 21) & (df['weekly_work_hours'] <= 30), 'wwh_group'] = '21-30'
df.loc[df['weekly_work_hours'] > 30, 'wwh_group'] = 'more than 30'
df['wwh_group'].value_counts()
```

7- Build a risk score by summing:
2 points if the tuition is not up to date;
1 point if stress_level is at least 8;
1 point if sleep_hours is smaller than 5;
1 point if the student has a part-time job with more than 20 weekly work hours;
2 points if attendance_rate is smaller than 60;
1 point if missed_deadlines is larger than 3.

For each value of the risk score, compute the number of students and the dropout rate.

```python
df['risk_score'] = 0
df.loc[df['tuition_up_to_date'] == 0, 'risk_score'] += 2
df.loc[df['stress_level'] >= 8, 'risk_score'] += 1
df.loc[df['sleep_hours'] < 5, 'risk_score'] += 1
df.loc[(df['part_time_job'] == 1) & (df['weekly_work_hours'] > 20), 'risk_score'] += 1
df.loc[df['attendance_rate'] < 60, 'risk_score'] += 2
df.loc[df['missed_deadlines'] > 3, 'risk_score'] += 1
count_risk = df.groupby('risk_score')['person_id'].count()
dropout_rate = df.groupby('risk_score')['dropout'].mean()
student_risk_score = pd.DataFrame({'number_of_students' : count_risk, 'dropout_rate' : dropout_rate})
student_risk_score
```

8- Read the GDP per capita dataset published by Our World in Data directly from the URL

https://ourworldindata.org/grapher/gdp-per-capita-worldbank.csv?v=1&csvType=full&useColumnShortNames=true
and store it in a dataframe called gdp. The column ny_gdp_pcap_pp_kd contains the GDP per capita.

Clean the dataset, removing the rows that do not correspond to a single country (e.g. World, High-income countries). For each country keep only the row with the most recent year.

```python
# AI disclosure: the storage_options parameter (User-Agent header) was obtained via Gemini to resolve an HTTP 403 error
gdp = pd.read_csv('https://ourworldindata.org/grapher/gdp-per-capita-worldbank.csv?v=1&csvType=full&useColumnShortNames=true', storage_options={'User-Agent': 'Our World In Data data fetch/1.0'})
gdp.head()
```

```python
gdp_fin = gdp[gdp['code'].str.len() == 3]
idx = gdp_fin.groupby('code')['year'].idxmax()
gdp = gdp_fin.loc[idx]
gdp
```
